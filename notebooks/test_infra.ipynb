```python
spark.sql("""
CREATE NAMESPACE IF NOT EXISTS medical.sandbox
""")

spark.sql("SHOW NAMESPACES IN medical").show()
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS medical.sandbox.patients_test (
    patient_id BIGINT,
    age INT
)
USING iceberg
""")

spark.sql("""
INSERT INTO medical.sandbox.patients_test
VALUES (1, 45), (2, 32)
""")

spark.sql("""
SELECT * FROM medical.sandbox.patients_test
""").show()
```

```python
spark.sql("""
SELECT
    snapshot_id,
    committed_at,
    operation
FROM medical.sandbox.patients_test.snapshots
""").show(truncate=False)
```
